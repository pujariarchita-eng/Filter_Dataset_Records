### Dataset
Student Performance Dataset

```python
import pandas as pd

print("Pandas imported successfully!")
```

```python
data = {
    "Student_ID": range(1, 16),
    "Name": [
        "Aarav", "Priya", "Rahul", "Sneha", "Arjun",
        "Ananya", "Rohan", "Kavya", "Vikram", "Neha",
        "Aditya", "Pooja", "Karan", "Isha", "Manish"
    ],
    "Gender": [
        "Male", "Female", "Male", "Female", "Male",
        "Female", "Male", "Female", "Male", "Female",
        "Male", "Female", "Male", "Female", "Male"
    ],
    "Age": [
        20, 21, 19, 22, 20,
        21, 23, 20, 22, 19,
        21, 22, 20, 23, 21
    ],
    "Math_Score": [
        85, 92, 67, 78, 95,
        88, 72, 90, 65, 81,
        89, 76, 94, 70, 83
    ],
    "Science_Score": [
        82, 95, 70, 80, 91,
        85, 75, 93, 68, 79,
        87, 74, 96, 72, 84
    ],
    "Attendance": [
        92, 95, 78, 85, 98,
        91, 82, 96, 75, 88,
        94, 80, 97, 83, 90
    ]
}

df = pd.DataFrame(data)

df
```

```python
df["Math_Score"] > 80
```

```python
high_attendance = df[df["Attendance"] > 90]

print("Students with Attendance above 90:")
display(high_attendance)

print("Number of records:", len(high_attendance))
```

```python
high_performers = df[
    (df["Math_Score"] > 80) &
    (df["Attendance"] > 90)
]

print("Students with Math Score above 80 AND Attendance above 90:")
display(high_performers)

print("Number of records:", len(high_performers))
```

```python
excellent_students = df[
    (df["Math_Score"] > 90) |
    (df["Science_Score"] > 90)
]

print("Students with Math Score above 90 OR Science Score above 90:")
display(excellent_students)

print("Number of records:", len(excellent_students))
```

```python
female_high_attendance = df[
    (df["Gender"] == "Female") &
    (df["Attendance"] > 85)
]

print("Female students with Attendance above 85:")
display(female_high_attendance)

print("Number of records:", len(female_high_attendance))
```

```python
df["Gender"] == "Female"
```

```python
df["Attendance"] > 85
```

```python
medium_math = df[
    (df["Math_Score"] >= 70) &
    (df["Math_Score"] <= 90)
]

print("Students with Math Score between 70 and 90:")
display(medium_math)

print("Number of records:", len(medium_math))
```

```python
medium_math = df[
    (df["Math_Score"] >= 70) &
    (df["Math_Score"] <= 90)
]

print("Students with Math Score between 70 and 90:")
display(medium_math)

print("Number of records:", len(medium_math))
```

```python
print("Original number of records:", len(df))
print("Attendance > 90:", len(high_attendance))
print("Math > 80 AND Attendance > 90:", len(high_performers))
print("Math > 90 OR Science > 90:", len(excellent_students))
print("Female AND Attendance > 85:", len(female_high_attendance))
print("Math Score between 70 and 90:", len(medium_math))
```
